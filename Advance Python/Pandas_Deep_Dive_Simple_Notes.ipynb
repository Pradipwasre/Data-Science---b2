# 1. What is a DataFrame?

A **DataFrame** is like a table.

It has:
- rows
- columns
- an index

Think:

`DataFrame = rows + columns + index`

### Example 1
Create a student table with a custom index: A, B, C, D, E.

### Example 2
Create a small sales table with a normal index.

````python
# Example 1: Student DataFrame

df = pd.DataFrame(
    {
        "Name": ["Amit", "Neha", "Raj", "Priya", "Karan"],
        "Age": [22, 24, 21, 25, 23],
        "City": ["Pune", "Mumbai", "Nagpur", "Delhi", "Nashik"]
    },
    index=["A", "B", "C", "D", "E"]
)

print(df)
````

Positions start from `0`.

So:

| Label | Position |

| A | 0 |
| B | 1 |
| C | 2 |
| D | 3 |

### Example 1

````python
# Example 1
print(df.iloc[2])
````

````python
# Example 2
print(df.iloc[3, 0])
````

### Example 1

### Example 2

````python
# Example 1
print(df.iloc[1:4])
````

````python
# Example 2
print(df.iloc[0:3, 0:2])
````

# 6. `.loc[]` vs `.iloc[]`

This is one of the most important Pandas ideas.

### `.loc[]`
Uses **labels**.

```python
df.loc["C"]
```

means:

> Find the row whose label is `"C"`.

### `.iloc[]`
Uses **positions**.

```python
df.iloc[2]
```

means:

> Find the row at position `2`.

Both can return the same row here, but they use different ways to find it.

### Example 1
Compare `.loc["C"]` and `.iloc[2]`.

### Example 2
Compare selecting `Name` from row D.

````python
# Example 1
print("Using loc:")
print(df.loc["C"])

print("\nUsing iloc:")
print(df.iloc[2])
````

````python
# Example 2
print("Using loc:")
print(df.loc["D", "Name"])

print("\nUsing iloc:")
print(df.iloc[3, 0])
````

## A very important experiment

Now change the row labels.

The **data does not move**, but the labels change.

This shows why `.loc` is about labels and `.iloc` is about positions.

### Example 1
Create a DataFrame with names as the index.

### Example 2
Use `.loc` and `.iloc` on it.

````python
# Example 1

people = pd.DataFrame(
    {
        "Age": [22, 24, 21],
        "City": ["Pune", "Mumbai", "Delhi"]
    },
    index=["Amit", "Neha", "Raj"]
)

print(people)
````

````python
# Example 2

print("loc -> label 'Neha'")
print(people.loc["Neha"])

print("\niloc -> position 1")
print(people.iloc[1])
````

# 7. Filtering

Filtering means:

> Keep only the rows that satisfy a condition.

For example:

```python
df["Age"] > 22
```

creates a Boolean result:

`True` or `False`

Then Pandas uses that Boolean result to select rows.

### Example 1
Find students older than 22.

### Example 2
Find students from Pune.

````python
# Example 1
print(df[df["Age"] > 22])
````

````python
# Example 2
print(df[df["City"] == "Pune"])
````

## Filtering with two conditions

We can combine conditions.

Use:
- `&` for AND
- `|` for OR

Put each condition inside parentheses.

### Example 1
Age greater than 22 AND city is Delhi.

### Example 2
Age less than 23 OR city is Mumbai.

````python
# Example 1
result = df[(df["Age"] > 22) & (df["City"] == "Delhi")]
print(result)
````

````python
# Example 2
result = df[(df["Age"] < 23) | (df["City"] == "Mumbai")]
print(result)
````

# 10. Missing Values

Real-world data is often incomplete.

For example, a student's city may be missing.

Pandas usually represents missing data using `NaN` (or `None` in some situations).

### Example 1
Create one missing city.

### Example 2
Create two missing ages.

````python
# Example 1

df_missing = df.copy()
df_missing.loc["C", "City"] = None

print(df_missing)
````

````python
# Example 2

df_missing2 = df.copy()
df_missing2.loc["B", "Age"] = None
df_missing2.loc["D", "Age"] = None

print(df_missing2)
````

## Checking missing values

`isnull()` asks:

> Is this value missing?

The answer is `True` or `False`.

### Example 1
See every missing-value position.

### Example 2
Count missing values in every column.

````python
# Example 1
print(df_missing.isnull())
````

````python
# Example 2
print(df_missing.isnull().sum())
````

## Removing missing values

`dropna()` removes rows that contain missing values.

### Example 1
Remove rows with missing values.

### Example 2
Remove rows only when `Age` is missing.

````python
# Example 1
print(df_missing.dropna())
````

````python
# Example 2
print(df_missing2.dropna(subset=["Age"]))
````

## Filling missing values

Instead of deleting data, we can replace missing values.

`fillna()` means:

> Fill the missing values.

### Example 1
Fill missing city with `"Unknown"`.

### Example 2
Fill missing age with the average age.

````python
# Example 1
filled = df_missing.fillna({"City": "Unknown"})
print(filled)
````

````python
# Example 2
filled = df_missing2.copy()
average_age = filled["Age"].mean()
filled["Age"] = filled["Age"].fillna(average_age)

print(filled)
````

# 13. Reading a CSV with a Different Delimiter

Not every file uses commas.

For example, this file uses `|`:

```text
Name|Age|City
Amit|22|Pune
Neha|24|Mumbai
```

We can tell Pandas which delimiter is being used.

### Example 1
Create a pipe-separated file.

### Example 2
Read it using `sep="|"`.

````python
# Example 1

with open("students_pipe.csv", "w") as file:
    file.write("Name|Age|City\n")
    file.write("Amit|22|Pune\n")
    file.write("Neha|24|Mumbai\n")

print("Pipe-separated CSV created")
````

````python
# Example 2

pipe_df = pd.read_csv("students_pipe.csv", sep="|")
print(pipe_df)
````

# 15. How `.loc[]` Works — Simple Internal Idea

When we write:

```python

```

we are saying:

> Find the row whose **label** is `"C"`.

The important thing is the **label**, not its position.

Change the labels and see that `.loc` follows the labels.

````python
# Example 1
print(df.loc["C"])
````

````python
# Example 2

demo = df.copy()
demo.index = ["X", "Y", "Z", "P", "Q"]

print(demo)

# The third row is now labeled Z
print("\nUsing loc with label Z:")
print(demo.loc["Z"])
````

# 16. How `.iloc[]` Works — Simple Internal Idea

When we write:

```python
df.iloc[2]
```

we are saying:

> Give me the row at integer position `2`.

Position starts at zero.

So:

`0 → first`

`1 → second`

`2 → third`

### Example 1
Get the third row.

### Example 2
Change the labels and prove that `.iloc` still uses position.

````python
# Example 1
print(df.iloc[2])
````

````python
# Example 2

demo = df.copy()
demo.index = ["X", "Y", "Z", "P", "Q"]

print("Third row by position:")
print(demo.iloc[2])
````

# 17. Boolean Mask — What Filtering Really Does

When we write:

```python
df["Age"] > 22
```

Pandas first creates a Boolean Series.

Example idea:

```text
A → False
B → True
C → False
D → True
E → True
```

Then Pandas keeps the rows where the answer is `True`.

### Example 1
Create the Boolean mask.

### Example 2
Use that mask to filter the DataFrame.

````python
# Example 1
mask = df["Age"] > 22
print(mask)
````

````python
# Example 2
mask = df["Age"] > 22
filtered_df = df[mask]

print(filtered_df)
````
