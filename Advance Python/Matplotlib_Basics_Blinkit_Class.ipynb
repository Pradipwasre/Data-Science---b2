In this notebook, we will learn the basics of **Matplotlib** using one small dataset.

### What we will learn
1. Create a DataFrame
2. Line Plot
3. Bar Plot
4. Histogram
5. Box Plot
6. Scatter Plot
7. Basic plot customization
8. Save a plot using `savefig()`

> **Learning goal:** Understand *when to use which plot* and how to create it with simple Python code.

## 1. Import Libraries

We need:
- `pandas` → to work with the dataset
- `matplotlib.pyplot` → to create graphs

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Create the Sample Dataset

Our dataset has:
- **City** → categorical data
- **Orders** → numerical data
- **Revenue** → numerical data

We will use the same dataset for all the plots.

```python
data = {
    "City": ["Pune", "Mumbai", "Nagpur", "Nashik", "Aurangabad", "Amravati"],
    "Orders": [120, 180, 90, 140, 110, 70],
    "Revenue": [2400, 3600, 1800, 2800, 2200, 1400]
}

df = pd.DataFrame(data)

df
```

## 3. Line Plot

A **line plot** is useful for seeing a trend or change across an ordered sequence.

Here, we are simply using cities on the x-axis and revenue on the y-axis.

```python
plt.plot(df["City"], df["Revenue"])

plt.title("Blinkit Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.show()
```

### 💡 Remember
**Line Plot → Trend / change**

## 4. Bar Plot

A **bar plot** is useful for comparing different categories.

Here, we compare the number of orders for each city.

```python
plt.bar(df["City"], df["Orders"])

plt.title("Orders by City")
plt.xlabel("City")
plt.ylabel("Orders")

plt.show()
```

### 💡 Remember
**Bar Plot → Category comparison**

## 5. Histogram

A **histogram** shows the distribution of numerical data.

Here, we look at how the revenue values are distributed.

```python
plt.hist(df["Revenue"])

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

### 💡 Remember
**Histogram → Distribution of numerical data**

## 6. Box Plot

A **box plot** helps us understand the spread of numerical data and can also help identify outliers.

Here, we look at the spread of revenue.

```python
plt.boxplot(df["Revenue"])

plt.title("Revenue Spread")
plt.ylabel("Revenue")

plt.show()
```

### 💡 Remember
**Box Plot → Spread + possible outliers**

## 7. Scatter Plot

A **scatter plot** is useful for seeing the relationship between two numerical variables.

Here, we compare **Orders** and **Revenue**.

```python
plt.scatter(df["Orders"], df["Revenue"])

plt.title("Orders vs Revenue")
plt.xlabel("Orders")
plt.ylabel("Revenue")

plt.show()
```

### 💡 Remember
**Scatter Plot → Relationship between two numerical variables**

## 8. Basic Plot Customization

We can change:
- `color` → color of the line
- `marker` → points on the line
- `linestyle` → style of the line
- `linewidth` → thickness of the line
- `label` → name shown in the legend

We will customize the same revenue line plot.

```python
plt.plot(
    df["City"],
    df["Revenue"],
    color="green",
    marker="o",
    linestyle="--",
    linewidth=2,
    label="Revenue"
)

plt.title("Blinkit Revenue Analysis")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.legend()

plt.show()
```

## 9. Save a Plot

Use `plt.savefig()` to save a figure as an image file.

The file will be saved in the notebook's current folder.

```python
plt.bar(df["City"], df["Orders"])

plt.title("Orders by City")
plt.xlabel("City")
plt.ylabel("Orders")

plt.savefig("orders_plot.png")
plt.show()
```

## 🧠 Quick Plot Selection Guide

| Plot | Main Use |
|---|---|
| **Line Plot** | Trend / change |
| **Bar Plot** | Compare categories |
| **Histogram** | Distribution |
| **Box Plot** | Spread / outliers |
| **Scatter Plot** | Relationship between two numerical variables |

### Simple rule

**Category comparison → Bar**  
**Distribution → Histogram**  
**Spread / outliers → Box**  
**Two numerical variables → Scatter**  
**Trend → Line**
