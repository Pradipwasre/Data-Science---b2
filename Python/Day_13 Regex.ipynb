# Regex in Python

## What is Regex?
Regex (short for **Regular Expression**) is a way to **search, match, or manipulate text** using patterns.  
Think of it like a "smart search tool" that can find words, numbers, or specific formats inside strings.

## Simple Example
```python
import re

# Example string
text = "My phone number is 12345"

# Search for digits
pattern = r"\d+"   # \d means any digit, + means one or more

result = re.findall(pattern, text)
print(result)   # Output: ['12345']

-- Here, \d+ finds all numbers in the text.


````python
import re
text = "my phone number is 9988776655"

# serch for digits 
pattern = r"\d+"

result = re.findall(pattern , text)
print(result)
````

# Common Regex Notations in Python

- **\d** → Matches any digit (0–9)  
- **\D** → Matches any non-digit character  
- **\w** → Matches any word character (letters, digits, underscore)  
- **\W** → Matches any non-word character  
- **\s** → Matches any whitespace (space, tab, newline)  
- **\S** → Matches any non-whitespace character  
- **.** → Matches any single character (except newline)  
- **^** → Matches the start of a string  
- **$** → Matches the end of a string  
- **+** → Matches one or more repetitions  
- ***** → Matches zero or more repetitions  
- **?** → Matches zero or one occurrence  
- **{n}** → Matches exactly *n* repetitions  
- **{n,m}** → Matches between *n* and *m* repetitions  
- **|** → OR operator (choice between patterns)  
- **[]** → Matches any single character inside the brackets  
- **[^ ]** → Matches any character *not* inside the brackets

````python
# \D → Matches any **non-digit character** (anything that is not 0–9).

text = "code123!"

# make a pattern
pattern = r"\D+"

result = re.findall(pattern , text)
print(result)
````

````python
# \w → Matches any **word character** (letters, digits, underscore).

text = "python_3 is very powerful tool!"

# Make a pattern
pattern = r"\w+"

result = re.findall(pattern , text)
print(result)
````

````python
# **\W** → Matches any **non-word character** (anything that is NOT a letter, digit, or underscore).

text = "Python_3 is great!!!"

pattern = f"\W+"  ## \W means non-word character, + means one or more

result = re.findall(pattern , text)
print(result)
````

````python
# **\s** → Matches any **whitespace character** (space, tab, newline).

text = "Python is fun to learn"

pattern = r"\s" # \s means whitespace, + means one or more

result = re.findall(pattern, text) 
print(result)
````

````python
# **\S** → Matches any **non-whitespace character** (anything that is not a space, tab
text = "python is fun to learn"

pattern = r"\S+" # \S means non-whitespace, + means one or more

result= re.findall(pattern , text)
print(result)
````

````python
# **.** → Matches any **single character** (except newline `\n`).

text = "Data science is fun"

pattern = r"."

result = re.findall(pattern , text)
print(result)
````

````python
# **^** → Matches the **start of a string** (or start of a line in multiline mode).

text = "Python is powerful and fun"

# Match word only at the start
pattern = r"^Python"

result = re.findall(pattern, text)
print(result)
````

````python
# **$** → Matches the **end of a string** (or end of a line in multiline mode).

text = "Regex is powerful tool"

# Match word only at the end
pattern = r"tool$"   # $ ensures match at the end

result = re.findall(pattern, text)
print(result)   
````

````python
# + → Matches **one or more repetitions** of the preceding character or pattern.

text = "rrr rocks really right!"

pattern = r"r+"

result = re.findall(pattern , text)
print(result)
````
