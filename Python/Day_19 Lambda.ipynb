# Lambda Funtions

- write in single line 
- when we wanted a quick code 
- for the quick check 

```python
def add(a, b ):
    return a + b 
```

```python
add = lambda a , b : a + b
```

```python
mul = lambda a, b : a * b 
```

```python
add(19 , 20)
```

```python
mul(10,10)
```

```python
# write a function for the odd even using the def function 
```

```python
def num_check(num):
    if num % 2 == 0:
        print("even")
    else:
        print("odd")
```

```python
num_check(1)
```

```python
num_check(10)
```

```python
num_check = lambda num : "even" if num % 2 == 0 else "Odd"
```

```python
num_check(1)
```

```python
# voting card : "can apply" & "can't apply" 
voting_card = lambda age: "can Apply" if age >= 18 else "Can't apply" 
```
