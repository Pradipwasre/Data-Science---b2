# 🔄 Looping Logic – Mastering Iteration## Why Do We Need Loops?Imagine you need to perform the same task multiple times — like greeting 100 students, adding numbers from 1 to 1000, or checking each item in a shopping list. Writing the same code over and over is boring, inefficient, and error-prone.**Loops** solve this problem by letting us write a block of code once and run it as many times as we need. They are one of the most powerful tools in programming.## Types of Loops in Python| Loop Type | When to Use ||-----------|-------------|| **for loop** | When you know exactly how many times to repeat (or when iterating over a collection) || **while loop** | When you want to repeat until a condition becomes false || **break** | To exit a loop early || **continue** | To skip the current iteration and move to the next || **pass** | To write an empty loop (placeholder for future code) |---## for LoopA `for` loop goes through each item in a sequence (like a list, string, or range) one by one.**Syntax:**```pythonfor item in sequence:    # do something with item```## while LoopA `while` loop keeps running as long as its condition stays `True`.**Syntax:**```pythonwhile condition:    # do something```## Loop Control Statements- **break** — Stops the loop immediately- **continue** — Skips to the next iteration- **pass** — Does nothing (used when code is required syntactically)

### for Loop – Example 1: Printing "Hello" 100 Times

````python
greetings = []for i in range(100):    greetings.append("Hello")greetings[:5]  # showing first 5 items
````

### for Loop – Example 2: Iterating Over a List

````python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]upper_fruits = []for fruit in fruits:    upper_fruits.append(fruit.upper())upper_fruits
````

### for Loop – Example 3: Squaring Numbers 1 to 10

````python
squares = []for num in range(1, 11):    squares.append(num * num)squares
````

### for Loop – Example 4: Iterating Over a String

````python
word = "Python"letters = []for letter in word:    letters.append(letter)letters
````

### for Loop – Example 5: Using enumerate() for Index and Value

````python
colors = ["red", "green", "blue"]indexed_colors = []for index, color in enumerate(colors):    indexed_colors.append((index, color))indexed_colors
````

---## while Loop

### while Loop – Example 1: Adding 1 to x Five Times

````python
x = 0count = 0while count < 5:    x = x + 1    count = count + 1x
````

### while Loop – Example 2: Countdown from 5 to 1

````python
countdown = []n = 5while n > 0:    countdown.append(n)    n = n - 1countdown
````

### while Loop – Example 3: Doubling Until Value Exceeds 100

````python
values = []value = 1while value <= 100:    values.append(value)    value = value * 2values
````

### while Loop – Example 4: Collecting Even Numbers Below 20

````python
evens = []n = 0while n < 20:    evens.append(n)    n = n + 2evens
````

### while Loop – Example 5: Simulating 3 Password Attempts

````python
password_attempts = []attempt = 1while attempt <= 3:    password_attempts.append(attempt)    attempt = attempt + 1password_attempts
````

---## break, continue, pass

### break – Example 1: Stop Loop When i Reaches 6

````python
numbers = []for i in range(1, 20):    if i == 6:        break    numbers.append(i)numbers
````

### continue – Example 2: Skip Number 5

````python
result = []for i in range(1, 11):    if i == 5:        continue    result.append(i)result
````

### pass – Example 3: Placeholder That Does Nothing

````python
items = ["a", "b", "c", "d"]processed = []for item in items:    if item == "c":        pass    processed.append(item)processed
````

### break – Example 4: Exit while Loop Early

````python
values = []n = 1while n <= 10:    if n == 4:        break    values.append(n)    n = n + 1values
````

### continue – Example 5: Skip Even Numbers in while Loop

````python
result = []n = 0while n < 10:    n = n + 1    if n % 2 == 0:        continue    result.append(n)result
````
