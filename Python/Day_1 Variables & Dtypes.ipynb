# Variables in Python

A **variable** is a name that stores data in your program.  
Think of it as a label you attach to a value so you can reuse it later.


## Key Points
- A variable is a name that refers to a value.
- You can store numbers, text, or other data types.
- Use the `=` (assignment operator) to assign a value.

# Data Types in Python

A **data type** defines the kind of value a variable can hold.  
Python has several built-in data types that help us work with different kinds of information.

---

## Common Data Types

- **Numeric Types**
  - `int` → whole numbers (e.g., `10`, `-5`)
  - `float` → decimal numbers (e.g., `3.14`, `-0.5`)
  - `complex` → numbers with real and imaginary parts (e.g., `2 + 3j`)

- **Sequence Types**
  - `str` → text (e.g., `"Hello"`)
  - `list` → ordered, changeable collection (e.g., `[1, 2, 3]`)
  - `tuple` → ordered, unchangeable collection (e.g., `(1, 2, 3)`)

- **Mapping Type**
  - `dict` → key-value pairs (e.g., `{"name": "Pradip", "age": 25}`)

- **Set Types**
  - `set` → unordered collection of unique items (e.g., `{1, 2, 3}`)
  - `frozenset` → immutable version of a set

- **Boolean Type**
  - `bool` → logical values `True` or `False`
