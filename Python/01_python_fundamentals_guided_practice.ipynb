## 1. Lists

A list is an **ordered, mutable** collection. It allows duplicates. Indexing starts at `0`.

- `append(x)`: add one item at the end
- `extend(items)`: add each item from an iterable
- `insert(i, x)`: insert at index `i`
- `remove(x)`: remove the first matching value; raises `ValueError` if missing
- `pop(i)`: remove and return an item; without `i`, removes the last item
- `clear()`: remove all items
- `index(x)`: index of the first match; raises `ValueError` if missing
- `count(x)`: number of occurrences
- `sort()`: sort the original list; `reverse()` reverses its current order
- `copy()`: shallow copy; `len()`, `min()`, `max()`, `sum()` are built-ins

```python
numbers = [3, 1, 4, 1]
numbers.append(5)             # Add one item
numbers.extend([6, 7])        # Add multiple items
numbers.insert(1, 2)          # Insert 2 at index 1
numbers
```

```python
items = [10, 20, 30, 20]
items.remove(20)               # Removes the first 20
removed_item = items.pop()     # Removes and returns the last item
items, removed_item
```

```python
names = ["Asha", "Ravi", "Asha", "Mina"]
first_asha = names.index("Asha")
asha_count = names.count("Asha")
number_of_names = len(names)
first_asha, asha_count, number_of_names
```

```python
scores = [72, 55, 91, 84]
scores.sort()                  # Changes scores in place
sorted_copy = scores.copy()
scores.reverse()               # Reverses current order; does not sort
minimum = min(scores)
maximum = max(scores)
total = sum(scores)
scores, sorted_copy, minimum, maximum, total
```

```python
temporary = [1, 2, 3]
temporary.clear()
temporary
```

## 2. Tuples

A tuple is **ordered and immutable**: its items cannot be reassigned after creation. It allows duplicates and is useful for fixed collections. Available methods include `count()` and `index()`. Built-ins `len()`, `min()`, `max()`, and `sum()` work when appropriate for the values.

```python
temperatures = (22, 25, 22, 19, 28)
temperatures.count(22), temperatures.index(25), len(temperatures), min(temperatures), max(temperatures), sum(temperatures)
```

You can read a tuple item, but assigning to it raises `TypeError`. To create a changed version, build a new tuple.

```python
point = (4, 7)
updated_point = (9, point[1])
updated_point
```

## 3. Sets

A set stores **unique** values and does not provide positional indexing. Use `set()` for an empty set; `{}` creates an empty dictionary.

- `add(x)` adds one value; `update(items)` adds many
- `remove(x)` raises `KeyError` if absent; `discard(x)` does not
- `pop()` removes and returns an arbitrary item; `clear()` empties the set; `copy()` makes a shallow copy
- `union()` combines unique values; `intersection()` finds shared values
- `difference()` finds values only in the left set
- `symmetric_difference()` finds values in either set but not both
- `issubset()` and `issuperset()` test set relationships

```python
skills = {"Python", "SQL"}
skills.add("Pandas")
skills.update(["NumPy", "Python"])  # Duplicate values are ignored
skills.discard("R")                 # Safe when R is absent
skills_copy = skills.copy()
skills, skills_copy
```

```python
A = {"Python", "SQL", "Git"}
B = {"Python", "Excel", "Git"}
A.union(B), A.intersection(B), A.difference(B), A.symmetric_difference(B)
```

```python
small_set = {1, 2}
large_set = {1, 2, 3}
small_set.issubset(large_set), large_set.issuperset(small_set)
```

## 4. Dictionaries (`dict`)

A dictionary stores **key-value pairs**. Keys are unique. Use it when values need labels, such as a person's name, age, or city.

- `keys()`, `values()`, `items()`: views of keys, values, and pairs
- `get(key, default)`: safely read a value; returns `None` if missing and no default is supplied
- `update(mapping)`: add or replace entries
- `pop(key)`: remove a named key and return its value
- `popitem()`: remove and return the last inserted pair
- `clear()`, `copy()`: empty or shallow-copy the dictionary
- `setdefault(key, default)`: return existing value or insert the default
- `dict.fromkeys(keys, value)`: create a dictionary with the given keys and initial value

```python
person = {"name": "Amit", "age": 22, "city": "Pune"}
list(person.keys()), list(person.values()), list(person.items())
```

```python
person.get("age"), person.get("role", "Not provided")
```

```python
profile = {"name": "Neha", "age": 24}
profile.update({"city": "Mumbai", "age": 25})
removed_age = profile.pop("age")
last_pair = profile.popitem()
profile, removed_age, last_pair
```

```python
student = {"name": "Raj"}
student.setdefault("name", "Unknown")  # Existing value remains
student.setdefault("city", "Nagpur")   # Missing key is inserted
subjects = dict.fromkeys(["Maths", "Python", "English"], 0)
student, subjects
```

## 5. Conditions

Conditions allow a program to choose which block runs. Python checks branches from top to bottom. Indentation is required.

- `if`: runs when its condition is true
- `else`: runs when earlier conditions are false
- `elif`: checks another condition if earlier conditions were false

```python
age = 18
if age >= 18:
    status = "Adult"
else:
    status = "Minor"
status
```

```python
num = int(input())

if num % 2 == 0:
    print("even")
else:
    print("Odd")
```

```python
marks = 76
if marks >= 90:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Needs improvement"
grade
```

## 6. Loops

Loops repeat code. A `for` loop visits items or a range; a `while` loop repeats while a condition is true.

`range(stop)` starts at `0` and stops before `stop`. `range(start, stop, step)` lets you choose the starting value and step.

- `break`: exit the nearest loop
- `continue`: skip to the next iteration
- `pass`: do nothing; acts as a placeholder

```python
squares = []
for number in range(1, 5):
    squares.append(number * number)
squares
```

```python
before_break = []
for number in range(1, 7):
    if number == 4:
        break
    before_break.append(number)

odd_numbers = []
for number in range(1, 7):
    if number % 2 == 0:
        continue
    odd_numbers.append(number)

if True:
    pass  # Placeholder for code to be added later

before_break, odd_numbers
```

```python
count = 1
visited = []
while count <= 3:
    visited.append(count)
    count += 1  # Update the condition variable so the loop ends
visited
```

## 7. Functions

A function is reusable code. `def` defines a function. **Parameters** are names in the definition; **arguments** are values passed when calling it. `return` sends a result back to the caller.

```python
def add(a, b):
    return a + b

result = add(4, 6)  # 4 and 6 are arguments
result
```

A parameter can have a default value. If the caller omits that argument, Python uses the default.

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

greet("Mina"), greet("Ravi", "Welcome")
```

A `lambda` creates a small anonymous function with one expression. Use `def` for more complex functions.

```python
square = lambda number: number ** 2
square(5)
```

`*args` collects extra positional arguments into a tuple. `**kwargs` collects extra keyword arguments into a dictionary. The names `args` and `kwargs` are conventions; the stars provide the behavior.

```python
def total(*args):
    return sum(args)

total(10, 20, 30)
```

```python
def describe_person(**kwargs):
    return kwargs

describe_person(name="Asha", city="Pune", age=23)
```

## 8. Exception handling

An exception is an error detected while code runs.

- `try`: code that may raise an exception
- `except`: handle a matching exception
- `else`: runs only if `try` succeeds
- `finally`: runs whether an exception occurred or not; useful for cleanup
- `raise`: deliberately signal an error

Prefer handling a specific exception rather than using a bare `except`, which can hide unrelated bugs.

```python
text_value = "42"
try:
    value = int(text_value)
except ValueError:
    result = "Please provide a whole number."
else:
    result = value * 2
finally:
    cleanup_message = "Conversion attempt finished."

result, cleanup_message
```

```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero."

divide(10, 0), divide(10, 2)
```

```python
def set_exam_score(score):
    if not 0 <= score <= 100:
        raise ValueError("Score must be between 0 and 100.")
    return score

set_exam_score(85)
```

| Topic | Main idea | Example syntax |

| List | Ordered, mutable collection | `items.append(value)` |

| Function | Reuse logic | `def calculate(...):` |
| Lambda | One-expression function | `lambda x: x * 2` |

| Exceptions | Handle or raise errors | `try / except / else / finally` |
