# Harnessing Lists - Functions, Comprehension, and Data Management

## Understanding Lists

A **list** in Python is a built-in data type used to store a collection of items in a single variable. Lists are one of the four main collection data types in Python, the others being tuples, sets, and dictionaries.

A list has the following important properties:

1. **Ordered** - The items in a list have a defined order, and this order will not change unless we change it ourselves. Each item has a fixed position called an index, starting from 0.
2. **Mutable** - Once a list is created, we can change, add, and remove items from it. This is different from a tuple, which cannot be changed after creation.
3. **Heterogeneous** - A single list can hold items of different data types together, such as integers, strings, floats, booleans, and even other lists.
4. **Allows Duplicates** - Since list items are indexed by position and not by value, a list can have the same value repeated multiple times.
5. **Dynamic in Size** - A list can grow or shrink at run time. We do not need to declare its size in advance, unlike arrays in some other languages.

### How a List Works Internally

Internally, a Python list is implemented as a dynamic array. This means Python keeps a block of memory that holds references (pointers) to the actual objects stored in the list, not the objects themselves. Because the list stores references rather than the raw values, it is possible for a single list to hold items of completely different types.

When we add more items than the current allocated space can hold, Python automatically allocates a larger block of memory, copies the existing references into it, and then adds the new item. This resizing happens behind the scenes, which is why lists appear to grow dynamically even though the underlying structure is a fixed-size array at any given moment.

### How to Create a List

A list is created by placing all the items inside square brackets `[ ]`, separated by commas. An empty list can also be created using `[]` or the `list()` constructor.

````python
empty_list = []
numbers = [10, 20, 30, 40, 50]
mixed_list = [1, 'apple', 3.5, True, ['nested', 'list']]

print('Empty list :', empty_list)
print('Numbers list :', numbers)
print('Mixed list :', mixed_list)
print('Type of numbers :', type(numbers))
````

## Indexing and Slicing in Lists

### Indexing

Every item in a list has a position number called an index. Indexing allows us to access a single item from the list using its position.

- **Positive indexing** starts from 0 for the first item, 1 for the second item, and so on.
- **Negative indexing** starts from -1 for the last item, -2 for the second last item, and so on. Negative indexing is useful when we want to access items from the end of the list without knowing its exact length.

### Slicing

Slicing allows us to access a range or a portion of the list instead of a single item. The general syntax is:

```
list_name[start : stop : step]
```

- `start` is the index where the slice begins (included).
- `stop` is the index where the slice ends (excluded).
- `step` decides how many items to skip between each selected item.

If `start` is left blank, slicing begins from the first item. If `stop` is left blank, slicing continues till the last item. If `step` is left blank, it defaults to 1.

````python
fruits = ['apple', 'banana', 'cherry', 'mango', 'orange', 'grapes']

print('Full list :', fruits)
print('First item (index 0) :', fruits[0])
print('Third item (index 2) :', fruits[2])
print('Last item (index -1) :', fruits[-1])
print('Second last item (index -2) :', fruits[-2])
````

````python
print('Slice from index 1 to 3 :', fruits[1:4])
print('Slice from start to index 2 :', fruits[:3])
print('Slice from index 3 to end :', fruits[3:])
print('Every alternate item :', fruits[::2])
print('Reversed list using slicing :', fruits[::-1])
````

## List Methods

Python provides many built-in methods to work with lists. These methods let us add, remove, search, count, and reorganize the items of a list very easily. Below, each method is explained in theory first and then demonstrated with an example.

### 1. append()

The `append()` method adds a single item to the end of the existing list. It always adds exactly one item, even if that item is itself a list, in which case the whole list is added as one nested item at the end.

````python
colors = ['red', 'green', 'blue']
print('Before append :', colors)

colors.append('yellow')
print('After append :', colors)
````

### 2. extend()

The `extend()` method adds multiple items to the end of a list at once. It takes an iterable, such as another list, and adds each of its items individually to the original list, unlike `append()` which would add the whole iterable as a single nested item.

````python
colors = ['red', 'green', 'blue']
more_colors = ['yellow', 'purple']
print('Before extend :', colors)

colors.extend(more_colors)
print('After extend :', colors)
````

### 3. insert()

The `insert()` method adds an item at a specific position in the list. It takes two arguments, the index at which the item should be placed and the item itself. All the items after that position are shifted one place to the right.

````python
colors = ['red', 'green', 'blue']
print('Before insert :', colors)

colors.insert(1, 'orange')
print('After insert at index 1 :', colors)
````

### 4. remove()

The `remove()` method deletes the first matching occurrence of a given value from the list. If the value is not present in the list, Python raises a `ValueError`.

````python
colors = ['red', 'green', 'blue', 'green']
print('Before remove :', colors)

colors.remove('green')
print('After remove (first green removed) :', colors)
````

### 5. pop()

The `pop()` method removes an item from the list and also returns that removed item, so it can be stored in a variable. If no index is given, it removes the last item by default. If an index is given, it removes the item at that specific position.

````python
colors = ['red', 'green', 'blue', 'yellow']
print('Before pop :', colors)

last_item = colors.pop()
print('Popped item (default, last) :', last_item)
print('List after pop() :', colors)

first_item = colors.pop(0)
print('Popped item (index 0) :', first_item)
print('List after pop(0) :', colors)
````

### 6. clear()

The `clear()` method removes all the items from the list, leaving it as an empty list. The list itself still exists, only its contents are erased.

````python
colors = ['red', 'green', 'blue']
print('Before clear :', colors)

colors.clear()
print('After clear :', colors)
````

### 7. index()

The `index()` method returns the position of the first occurrence of a given value in the list. If the value does not exist in the list, Python raises a `ValueError`.

````python
colors = ['red', 'green', 'blue', 'green']
position = colors.index('green')
print('List :', colors)
print('Index of first green :', position)
````

### 8. count()

The `count()` method returns the number of times a given value appears in the list. It is useful to check how many duplicates of a particular item exist.

````python
colors = ['red', 'green', 'blue', 'green', 'red', 'green']
green_count = colors.count('green')
print('List :', colors)
print('Count of green :', green_count)
````

### 9. sort()

The `sort()` method arranges the items of the list in place, meaning it changes the original list directly instead of returning a new one. By default it sorts in ascending order, but passing `reverse=True` sorts it in descending order.

````python
numbers = [40, 10, 30, 20, 50]
print('Before sort :', numbers)

numbers.sort()
print('After sort (ascending) :', numbers)

numbers.sort(reverse=True)
print('After sort (descending) :', numbers)
````

### 10. reverse()

The `reverse()` method flips the order of the items in the list in place. It does not sort the values, it simply reverses whatever order they were already in.

````python
numbers = [10, 20, 30, 40, 50]
print('Before reverse :', numbers)

numbers.reverse()
print('After reverse :', numbers)
````

## Example: Adding Grocery Items

To bring all these list methods together, here is a practical example of maintaining a grocery shopping list using a Python list. This example applies `append()`, `extend()`, `insert()`, `remove()`, `pop()`, `index()`, `count()`, `sort()`, `reverse()`, and finally `clear()` on the same grocery list, step by step.

````python
grocery_list = ['milk', 'bread', 'eggs']
print('Initial grocery list :', grocery_list)
````

````python
grocery_list.append('butter')
print('After adding butter with append() :', grocery_list)
````

````python
grocery_list.extend(['rice', 'sugar'])
print('After adding rice and sugar with extend() :', grocery_list)
````

````python
grocery_list.insert(2, 'cheese')
print('After inserting cheese at index 2 :', grocery_list)
````

````python
grocery_list.remove('bread')
print('After removing bread :', grocery_list)
````

````python
last_bought_item = grocery_list.pop()
print('Item removed with pop() :', last_bought_item)
print('Grocery list now :', grocery_list)
````

````python
cheese_position = grocery_list.index('cheese')
print('Position of cheese in the list :', cheese_position)
````

````python
grocery_list.append('milk')
milk_count = grocery_list.count('milk')
print('Grocery list :', grocery_list)
print('Number of times milk appears :', milk_count)
````

````python
grocery_list.sort()
print('Grocery list sorted alphabetically :', grocery_list)
````

````python
grocery_list.reverse()
print('Grocery list in reverse order :', grocery_list)
````

````python
grocery_list.clear()
print('Grocery list after clear() :', grocery_list)
````

## Summary

A list is an ordered, mutable, and heterogeneous collection that can grow or shrink dynamically. Indexing lets us pick out a single item using its position, while slicing lets us pull out a whole portion of the list at once. The built-in list methods, `append()`, `extend()`, `insert()`, `remove()`, `pop()`, `clear()`, `index()`, `count()`, `sort()`, and `reverse()`, give us complete control to add, remove, search, count, and reorganize items, as shown in the grocery list example above.
