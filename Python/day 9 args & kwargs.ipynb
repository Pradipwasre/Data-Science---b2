```python
# find the sum of numbers
def sum_elements(*nums):

    total = 0

    for num in nums:
        total += num

    return total
```

```python
sum_elements(10,20)
```

## pass the elements in the key-value way... ? 

-- **kwargs

```python
def show_kwargs(**kwargs):
    print(kwargs)
```

```python
show_kwargs(name = "rock" , age = 20)
```

def add(a, b) # fixed parameter pass

def add(*nums) # *args passing multiple arguements not fixed

def add(**kwargs) # we will pass arguments like key values pair

```python
# print all keyword argument
def show_kwargs(**kwargs):
    print(kwargs)
```

```python
show_kwargs(city = "pune" , rain = False)
```

```python
def student_info(**record): # pass : same way key and value 
    print(record)
```

```python
student_info(name = "Rock" , city = "NYC" , age = 20, Grade = "A", sport = "Fighting")
```

```python
# loop the key-value pairs

def loop_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(key, value)
```

```python
loop_kwargs(name = "rock" , city = "NYC")
```
