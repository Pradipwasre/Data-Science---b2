# Understanding Sets in Python

## Introduction
A **set** in Python is an unordered collection of unique elements.  
It is classified as a **dynamic data structure** because elements can be added or removed at runtime.  
At the same time, the **elements themselves must be immutable** (e.g., numbers, strings, tuples), ensuring stability inside the set.



## Characteristics of Sets
- **Unordered** → No fixed position of elements.
- **Unique elements** → Duplicates are automatically removed.
- **Mutable container** → You can add or remove elements.
- **Immutable members** → Elements must be immutable types.



## Creating Sets
```python
# Using curly braces
numbers = {1, 2, 3, 4}

# Using set() constructor
fruits = set(["apple", "banana", "cherry"])

# Empty set
empty_set = set()   # {} creates a dictionary, not a set

## Methods: 

````python
# 1 .add
students = {"Amit", "Sneha"}
````

````python
students.add("Rahul")
print(students)
````

````python
# 2. .update()
fruits = {"apple", "banana"}
````

````python
fruits.update(["cherry", "mango"])
print(fruits)
````

````python
# 3. .remove()
numbers = {1, 2, 3, 4}
````

````python
numbers.remove(3)
print(numbers)
````

# Classroom Attendance Tracker

## Scenario
Imagine you are a teacher keeping track of students who attended class.  
A **set** is perfect because:
- It avoids duplicate names.
- It allows easy addition/removal of students.
- It supports quick membership checks.


## Example Code

```python
# Initial empty set for attendance
attendance = set()

# List of students arriving one by one
students_arriving = ["Amit", "Sneha", "Rahul", "Sneha", "Priya"]

# Using a simple loop to add them into the set
for student in students_arriving:
    attendance.add(student)   # duplicates automatically ignored

print("Final Attendance:", attendance)

````python
# Initial empty set for attendance
attendance = set()

# List of students arriving one by one
students_arriving = ["Amit", "Sneha", "Rahul", "Sneha", "Priya"]

# Using a simple loop to add them into the set
for student in students_arriving:
    attendance.add(student)   # duplicates automatically ignored

print("Final Attendance:", attendance)
````

````python
# Suppose Rahul left early
attendance.remove("Rahul")

print("Updated Attendance:", attendance)
````
