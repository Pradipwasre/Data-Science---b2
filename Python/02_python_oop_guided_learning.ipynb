## 1. Class

A **class** is a blueprint for creating objects. It describes the data (attributes) and actions (methods) that objects of that type can have.

### Basic syntax
```python
class ClassName:
    pass
```

`pass` is a placeholder. It lets Python accept a class body that has not been implemented yet.

````python
class Student:
    pass

# Student is the class. We will create an object from it in the next section.
type(Student).__name__
````

## 2. Object

An **object** is an instance of a class. A class is the blueprint; an object is a specific item created from that blueprint. Different objects can hold different data.

````python
class Student:
    pass

student_one = Student()
student_two = Student()

type(student_one).__name__, student_one is student_two
````

`student_one` and `student_two` are two separate objects created from the same `Student` class. The expression `is` checks whether two references point to the exact same object.

## 3. Constructor: `__init__()`

`__init__()` is the initializer method. Python calls it automatically after a new instance is created. It is commonly used to set the object's initial attributes.

### General syntax
```python
class ClassName:
    def __init__(self, argument):
        self.attribute = argument
```

- `def` defines a method.
- `__init__` is the special initializer method.
- `self` refers to the current object.
- `argument` is the value supplied when creating the object.
- `self.attribute = argument` stores that value on the object.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student_one = Student("Asha", 20)
student_two = Student("Ravi", 22)

student_one.name, student_one.age, student_two.name, student_two.age
````

Notice that `Student("Asha", 20)` passes two arguments. Python supplies the new object as `self` automatically, so you do not write `self` when creating the object.

## 4. Understanding `self`

`self` is the conventional name for the first parameter of an instance method. It refers to the particular object on which the method is called. Python passes that object automatically when you call the method through the object.

For example, `student_one.show_details()` is conceptually similar to calling `Student.show_details(student_one)`. You normally use the first form.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        return f"Name: {self.name}, Age: {self.age}"

student_one = Student("Asha", 20)
student_two = Student("Ravi", 22)

student_one.show_details(), student_two.show_details()
````

The method reads `self.name` and `self.age`, so it uses the data belonging to the object that called it. `self` is not a Python keyword, but using this name is the standard convention.

## 5. Arguments and parameters

A **parameter** is a variable listed in a function or method definition. An **argument** is the actual value passed to the function or method when calling it.

In `def greet(self, name):`, `name` is a parameter. In `student.greet("Mina")`, `"Mina"` is an argument. For instance methods, Python supplies `self` automatically.

````python
class Greeter:
    def greet(self, name, greeting="Hello"):
        return f"{greeting}, {name}!"

greeter = Greeter()

# Positional arguments
a = greeter.greet("Mina", "Welcome")

# A keyword argument and a default argument
b = greeter.greet(name="Ravi")

a, b
````

### Interview note
- **Positional argument:** matched by its position.
- **Keyword argument:** matched by parameter name.
- **Default parameter:** uses a preset value when no argument is provided.

When calling a method, do not pass a value for `self` explicitly.

## 6. Methods

A **method** is a function defined inside a class. An instance method can read or update the object's attributes through `self`.

The example below includes one method that returns information and another that changes the object's state.

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def get_balance(self):
        return self.balance

account = BankAccount("Neha", 1000)
new_balance = account.deposit(250)
current_balance = account.get_balance()
new_balance, current_balance
````

`deposit()` changes the balance, so it changes the object's state. `get_balance()` reads the balance and returns it. In a real banking application, you would also validate amounts and control access to account data.

## 7. Inheritance

**Inheritance** allows a child class to reuse and extend attributes or methods from a parent class.

```python
class Child(Parent):
    pass
```

A child class can define its own methods and can override inherited methods. Use inheritance when the child genuinely represents a more specific kind of the parent.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof"

class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow"

dog = Dog("Bruno")
cat = Cat("Luna")

dog.name, dog.speak(), cat.name, cat.speak()
````

`Dog` and `Cat` inherit `__init__()` from `Animal`, so their objects receive a `name`. Each child class supplies its own version of `speak()`.

### Using `super()`
`super()` lets a child class call an implementation from its parent class. It is commonly used when a child needs to add setup while retaining the parent's setup.

````python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, employee_id):
        super().__init__(name)
        self.employee_id = employee_id

employee = Employee("Amit", "E104")
employee.name, employee.employee_id
````

Here, `super().__init__(name)` runs the `Person` initializer so `name` is set in one place. The `Employee` initializer then adds `employee_id`.

## 8. Polymorphism

**Polymorphism** means “many forms.” Different classes can provide the same method name but implement it differently. Code can call that method without needing to know the exact class of every object.

In Python, method overriding is a common way to achieve this. Python also supports flexible behavior through duck typing: if an object provides the method your code needs, the code can often use it.

````python
class EmailNotification:
    def send(self, message):
        return f"Email sent: {message}"

class SMSNotification:
    def send(self, message):
        return f"SMS sent: {message}"

class PushNotification:
    def send(self, message):
        return f"Push notification sent: {message}"

def notify(channel, message):
    return channel.send(message)

channels = [EmailNotification(), SMSNotification(), PushNotification()]
results = [notify(channel, "Your order is ready") for channel in channels]
results
````

The function `notify()` uses the same call, `channel.send(message)`, for all three objects. Each object's class determines what `send()` does.

### Interview distinction: overriding vs. overloading
- **Overriding:** a child class provides its own implementation of an inherited method.
- **Overloading:** in some languages, several methods share a name but have different parameter signatures. Python does not support traditional signature-based method overloading in the same way; common alternatives include default arguments, `*args`, or dispatch techniques.

## 9. Encapsulation

**Encapsulation** means grouping data and the methods that operate on it, while controlling how the data is accessed or changed.

Python uses naming conventions rather than strict access modifiers like `private` in some other languages:
- `name` is a public attribute.
- `_name` is a convention meaning “internal use; please treat as non-public.”
- `__name` triggers name mangling, which helps avoid accidental name conflicts. It does not make data absolutely inaccessible.

A good class validates changes through methods or properties instead of allowing invalid state.

````python
class BankAccount:
    def __init__(self, owner, opening_balance=0):
        if opening_balance < 0:
            raise ValueError("Opening balance cannot be negative.")
        self.owner = owner
        self.__balance = opening_balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be greater than zero.")
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be greater than zero.")
        if amount > self.__balance:
            raise ValueError("Insufficient funds.")
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

account = BankAccount("Priya", 1000)
account.deposit(300)
account.withdraw(150)
account.get_balance()
````

The balance is stored as `__balance` and updated through methods that validate the request. In Python, name mangling changes its internal name; it is not a security boundary. Encapsulation is mainly about a clear, controlled interface and preserving valid object state.

The following property example provides attribute-style reading while keeping the underlying attribute internal.

````python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def celsius(self):
        return self.__celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero.")
        self.__celsius = value

temperature = Temperature(25)
initial_temperature = temperature.celsius
temperature.celsius = 30
initial_temperature, temperature.celsius
````

The setter validates each assignment to `temperature.celsius`. The property allows the class to control updates without changing how callers read or assign the attribute.

## 10. Abstraction

**Abstraction** exposes what an object can do while hiding the details of how it does it. In Python, the `abc` module provides abstract base classes and abstract methods.

An abstract method describes an operation that a concrete child class must implement. You cannot instantiate a class while it still has unimplemented abstract methods.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        """Return the area of this shape."""
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

shapes = [Rectangle(4, 5), Circle(3)]
areas = [shape.area() for shape in shapes]
areas
````

`Shape` defines a common interface: every concrete shape must provide `area()`. `Rectangle` and `Circle` implement the calculation differently. The caller can use `area()` without knowing the details of each formula.

### Interview note: abstraction vs. encapsulation
- **Abstraction:** focuses on what operations are available and hides implementation complexity.
- **Encapsulation:** groups data and behavior and controls how internal state is accessed or changed.

They are related, but they solve different design problems.

## 11. Bringing the concepts together

This small example combines classes, objects, constructors, `self`, methods, inheritance, overriding (a form of polymorphism), and encapsulation.

````python
class Employee:
    def __init__(self, name, salary):
        if salary < 0:
            raise ValueError("Salary cannot be negative.")
        self.name = name
        self.__salary = salary

    def describe(self):
        return f"{self.name} is an employee."

    def get_salary(self):
        return self.__salary

class Developer(Employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language

    def describe(self):
        return f"{self.name} develops software using {self.language}."

class DataAnalyst(Employee):
    def __init__(self, name, salary, tool):
        super().__init__(name, salary)
        self.tool = tool

    def describe(self):
        return f"{self.name} analyzes data using {self.tool}."

team = [
    Developer("Asha", 70000, "Python"),
    DataAnalyst("Ravi", 65000, "Pandas")
]
team_descriptions = [member.describe() for member in team]
team_salaries = [member.get_salary() for member in team]
team_descriptions, team_salaries
````

| Abstraction | Defines essential operations while hiding implementation details | `ABC` and `@abstractmethod` |
