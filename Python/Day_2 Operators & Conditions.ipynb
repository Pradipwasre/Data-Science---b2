# Operators in Python

Operators are special symbols in Python that perform operations on values and variables. They are the building blocks of expressions and allow us to manipulate data.

---

## Types of Operators

- **Arithmetic Operators**  
  Perform mathematical calculations.  
  Examples:  
  - `+` (Addition) → `5 + 3 = 8`  
  - `-` (Subtraction) → `10 - 4 = 6`  
  - `*` (Multiplication) → `4 * 3 = 12`  
  - `/` (Division) → `7 / 2 = 3.5`  
  - `//` (Floor Division) → `7 // 2 = 3`  
  - `%` (Modulus) → `7 % 2 = 1`  
  - `**` (Exponentiation) → `2 ** 3 = 8`

- **Comparison Operators**  
  Compare values and return `True` or `False`.  
  Examples:  
  - `<` (Less than) → `3 < 5` → True  
  - `>` (Greater than) → `7 > 2` → True  
  - `<=` (Less than or equal to) → `4 <= 6` → True  
  - `>=` (Greater than or equal to) → `5 >= 5` → True  
  - `==` (Equal to) → `3 == 3` → True  
  - `!=` (Not equal to) → `3 != 5` → True

- **Logical Operators**  
  Combine conditional statements.  
  Examples:  
  - `and` → True if both conditions are true  
  - `or` → True if at least one condition is true  
  - `not` → Negates the condition

- **Assignment Operators**  
  Assign values to variables, sometimes with modification.  
  Examples:  
  - `=` → `x = 10`  
  - `+=` → `x += 5` (adds 5 to `x`)  
  - `-=` → `x -= 3` (subtracts 3 from `x`)  
  - `*=` → `x *= 2` (multiplies `x` by 2)  
  - `/=` → `x /= 4` (divides `x` by 4)

---

## Example

```python
a = 10
b = 3

print(a + b)   # Arithmetic → 13
print(a > b)   # Comparison → True
print(a and b) # Logical → 3 (since both are non-zero)

# The `if` Statement in Python

The `if` statement is used to make decisions in a program.  
It allows you to run certain code only when a condition is true.


## Key Points
- The condition inside `if` must evaluate to `True` or `False`.
- If the condition is `True`, the indented block of code runs.
- If the condition is `False`, the block is skipped.

## Example

```python
age = 18

if age >= 18:
    print("You can apply for Voting Card")

# The `if-else` Statement in Python

The `if-else` statement allows your program to choose between two paths:
- Run one block of code if the condition is `True`.
- Run another block if the condition is `False`.

---

## Key Points
- `if` checks the condition.
- If the condition is `True`, the `if` block runs.
- If the condition is `False`, the `else` block runs.

---

## Example

```python
age = 16

if age >= 18:
    print("You are an adult")
else:
    print("You are a minor")


# The `if-elif-else` Statement in Python

The `if-elif-else` statement allows your program to check **multiple conditions** in sequence.  
It runs the first block where the condition is `True`. If none are `True`, the `else` block runs.


## Key Points
- Start with `if` for the first condition.
- Use `elif` (else if) for additional conditions.
- Use `else` for the default case when none of the conditions are true.
- Only one block will execute.


## Example

```python
marks = 72

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
else:
    print("Grade: D")
