# Python Dictionary (dict)

## Introduction
A **Dictionary** in Python is a built-in data type used to store **key-value pairs**.  
It is **unordered, mutable, and indexed by keys**.

- Keys → Unique identifiers (immutable types like strings, numbers, tuples).
- Values → Data associated with keys (can be any type).



## Dictionary Characteristics
- **Unordered**: No fixed sequence (before Python 3.7, insertion order wasn’t preserved).
- **Mutable**: Can be changed after creation.
- **Indexed by keys**: Access values using keys, not positions.
- **Unique keys**: Duplicate keys are not allowed; latest value overwrites.

````python
# With values
student = {"name": "Rock", "age": 25, "course": "Data Science"}
````

````python
# Accessing the elements : 
student["name"]        # Output: "Pradip"
````

````python
student.get("age")     # Output: 25
````

````python
student.get("marks", 0) # Default value if key not found → 0
````

````python
# updating the dict 
student["age"] = 26          # Update existing key
````

````python
student["marks"] = 90        # Add new key-value pair
````

| **[Method](ca://s?q=Python_dict_methods)** | **Description** |
| --- | --- |
| ``dict.keys()`` | Returns all keys |
| ``dict.values()`` | Returns all values |
| ``dict.items()`` | Returns key-value pairs |
| ``dict.update({...})`` | Updates dictionary |
| ``dict.copy()`` | Shallow copy |

````python
# Sample dictionary
student = {"name": "Rock", "age": 25, "course": "Data Science"}
````

## Methods

````python
print(student.keys())
# Output: dict_keys(['name', 'age', 'course'])
````

````python
print(student.values())
# Output: dict_values(['Pradip', 25, 'Data Science'])
````

````python
print(student.items())
# Output: dict_items([('name', 'Pradip'), ('age', 25), ('course', 'Data Science')])
````

````python
student.update({"marks": 90, "age": 26})
print(student)
# Output: {'name': 'Pradip', 'age': 26, 'course': 'Data Science', 'marks': 90}
````

````python
student_copy = student.copy()
print(student_copy)
# Output: {'name': 'Pradip', 'age': 26, 'course': 'Data Science', 'marks': 90}
````

## Write a program to iterate through the dictionary and print only those students who scored above 50 marks.

````python
# Dictionary of students and marks
marks = {
    "Pradip": 45,
    "Gaurav": 78,
    "Sneha": 90,
    "Ravi": 32,
    "Meena": 55
}
````

````python
for name, score in marks.items():
    if score > 50:
        print(f"{name} → {score}")
````
